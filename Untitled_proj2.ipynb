```python
import joblib

model = joblib.load("Random Forest_best_model.pkl")  
encoder = joblib.load("label_encoder.pkl")
```

```python
import re
import string
import pandas as pd

# Suspicious keywords often seen in phishing or malicious URLs
suspicious_keywords = [
    'login', 'signin', 'verify', 'update',
    'banking', 'account', 'secure', 'ebay', 'paypal'
]

def extract_features(url: str) -> pd.Series:
    """
    Extracts a set of numerical features from a given URL.
    
    Parameters
    ----------
    url : str
        The URL to extract features from.
    
    Returns
    -------
    features : pd.Series
        A pandas Series containing the extracted features.
    """
    features = {}

    
    features['url_length'] = len(url)

    
    features['num_digits'] = sum(c.isdigit() for c in url)

    
    features['num_special_chars'] = sum(c in string.punctuation for c in url)

    
    features['num_subdomains'] = url.count('.') - 1

    
    features['has_ip'] = 1 if re.search(r'\d+\.\d+\.\d+\.\d+', url) else 0

    
    features['has_https'] = int(url.lower().startswith("https"))

    
    features['num_params'] = url.count('?')

    
    features['num_fragments'] = url.count('#')

    
    features['num_slashes'] = url.count('/')

    
    features['has_suspicious_words'] = int(any(word in url.lower() for word in suspicious_keywords))

    
    tld = url.split('.')[-1]
    features['tld_length'] = len(tld)

    
    features['is_common_tld'] = int(tld in ['com', 'org', 'net', 'edu', 'gov'])

    
    features['has_hex'] = int(bool(re.search(r'%[0-9a-fA-F]{2}', url)))

    
    features['repeated_chars'] = int(bool(re.search(r'(.)\1{3,}', url)))

    return pd.Series(features)
```

```python
test_urls = [
    "http://icicibank.com/privilege-banking/personal/Pure-Gold.page?utm_source=gold&utm_medium=text&utm_campaign=PB-flyout",           
    "http://www.wingert-sonnenschutz.de/index.php?option=com_content&view=article&id=52&Itemid=76",             
    "http://www.711100.com/cl/tpl/commonFile/css/jquery-ui/brown/jquery-ui-1.9.2.custom.css?v=ver17.32",          
    "www.osnews.com/story/5602"               
]

for url in test_urls:
    features = extract_features(url)
    features_df = pd.DataFrame([features])
    y_pred = model.predict(features_df)
    decoded_label = encoder.inverse_transform(y_pred)
    if decoded_label == 'benign':
        print("URL is: Safe")
    else:
        print(f"URL is: Unsafe ({decoded_label})")

```
